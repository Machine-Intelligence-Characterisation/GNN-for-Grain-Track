```python
from pathlib import Path
import numpy as np

type = "Compression"
# type = "Tension"

direction = "X"
direction = "XZ"
direction = "Z"

file_path = r'2D_10%strain_100grains\results_grain_matpoint'
base_path = Path(r"2D_10%strain_100grains")
file_path = base_path / "results_grain_matpoint" / f"{type}{direction}"

# Project root (GNN-for-Grain-Track)
PROJECT_ROOT = Path.cwd().parent
```

Read strain_average

```python
index = 0
measurement = "for_ML_strain_average"
loadcase = f"{type}{direction}"

filename = (
    PROJECT_ROOT
    / "2D_10%strain_100grains"
    / "results_grain_matpoint"
    / loadcase
    / measurement
    / f"grain_{index}.txt"
)

with open(filename, "r") as f:
    lines = [line.strip() for line in f if line.strip()]

headers = lines[:10] # First 10 lines are headers

data = {}
i = 10  # start after headers

while i < len(lines):
    label = lines[i]          # e.g. increment_0
    values = lines[i+1:i+10]  # next 9 lines
    matrix = np.array(values, dtype=float).reshape(3, 3)

    data[label] = matrix
    i += 10

print(headers)
print(data["increment_0"])
```
